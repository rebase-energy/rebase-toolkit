```python
import rebase as rb
import random
import math

def user_function(params):
    a = params["a"]
    b = params["b"]
    return a ** 2 + b ** 3

def benchmark_calculation(params):
    a = params["a"]
    b = params["b"]
    # GCD calculation (Euclidean algorithm)
    gcd = math.gcd(a, b)
    
    # LCM calculation
    lcm = abs(a * b) // gcd
    
    # Some additional complex operations (exponentiation)
    result = (a ** b) + (b ** a) - lcm
    
    return result

N = 10000
start=1
end=100
params = {
    "a": [random.randint(start, end) for _ in range(N)],
    "b": [random.randint(start, end) for _ in range(N)]
}

param_list = [dict(zip(params.keys(), values)) for values in zip(*params.values())]
```

```python
%%time
results = rb.sweep(benchmark_calculation, params)
print(results)
```

```python
%%time
results = []
for param in param_list: 
    result = benchmark_calculation(param)
    results.append(result)
print(results)
```
